```python
import sqlite3
import os 
```

```python
#Write a query to retrieve all records from the products table.
conn = sqlite3.connect("store.db") #Connect to the database
cursor = conn.cursor() #cursor is object that allow python to excute the sql query 
```

```python
ik=cursor.execute("""SELECT * FROM products""")
result = ik.fetchall()
print(result)
```

```python
for row in result:
    print(row)
```

```python
#Retrieve one product
cursor.execute("""
SELECT * FROM products WHERE id = 1""")


result = cursor.fetchone()

print(result)
```

```python
#Retrieve selected columns

cursor.execute("""

SELECT name , price , category FROM products 
WHERE id = 30 """)

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Search by category
cursor.execute(
    """SELECT * FROM products
    WHERE category = ?""" ,("honey",) #Places holders are passes as second arguments
)

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Task 5: Search using partial product names

#Like expect only one search problem at the time,

cursor.execute("""

SELECT * FROM products
WHERE name LIKE  ?""",("%honey%",))

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Retrieve all products sorted by price from lowest to highest.
cursor.execute("""
SELECT * FROM products
ORDER BY price ASC""")

result = cursor.fetchall()
for row in result:  
    print(row)
```

```python
#Task 8: Most expensive products first

cursor.execute(""" 
SELECT * FROM products
ORDER BY price DESC
""")

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Task 9: Find the most expensive product
cursor.execute(
    """
SELECT * FROM products
ORDER BY price DESC
LIMIT 1
    """
)

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Task 10: Find highly rated reviews

cursor.execute("""
SELECT id, rating, reviewer_name, 
review_text FROM reviews
WHERE rating >= 4""")

result = cursor.fetchall()
for row in result:
    print(row)
```

```python
#Task 11: Find the highest-rated review

cursor.execute("""

SELECT id, rating, reviewer_name, review_text FROM reviews
ORDER BY rating DESC
LIMIT 1""")

result = cursor.fetchall()
for row in result:  
    print(row)
```

```python
#"Show me all honey products costing less than Rs. 1,000, sorted from cheapest to most expensive."
cursor.execute("""
SELECT * FROM products 
WHERE category = ? AND price < ?""",("honey",1000))

result = cursor.fetchall()
for row in result:
    print(row)
```
