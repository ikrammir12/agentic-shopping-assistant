```python
from langgraph.graph import StateGraph,START ,END
#from typing import TypeDict
```

```python
#Create the state
class state():
    text:str
```

```python
#2=add the node
def clean_text(state: state):
    print("Node received :",state)
    return {"text":state["text"].strip().upper()}
```

```python
builder = StateGraph(state)
builder.add_node("clean_text",clean_text)
builder.add_edge(START,"clean_text")
builder.add_edge("clean_text",END)

app = builder.compile()
```

```python
from IPython.display import Image , display
display(Image(app.get_graph().draw_mermid()))
```
